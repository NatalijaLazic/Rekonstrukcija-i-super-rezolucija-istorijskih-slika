```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Putanje
TRAIN_DIR = '/content/drive/MyDrive/ML_projekat/dataset/DIV2K_train_HR/DIV2K_train_HR'
VALID_DIR = '/content/drive/MyDrive/ML_projekat/dataset/DIV2K_valid_HR/DIV2K_valid_HR'

# Učitaj listu fajlova
train_files = sorted([f for f in os.listdir(TRAIN_DIR) if f.endswith('.png')])
valid_files = sorted([f for f in os.listdir(VALID_DIR) if f.endswith('.png')])

print(f"Train slike: {len(train_files)}")
print(f"Valid slike: {len(valid_files)}")
print(f"\nPrvih 5 train fajlova: {train_files[:5]}")
print(f"Prvih 5 valid fajlova: {valid_files[:5]}")
```

```python
# Analiza dimenzija slika
print("Učitavamo informacije o slikama iz trening skupa:")

widths, heights = [], []

for fname in train_files:
    img = Image.open(os.path.join(TRAIN_DIR, fname))
    w, h = img.size
    widths.append(w)
    heights.append(h)

print(f"\n--- TRAIN SET ---")
print(f"Minimalna širina:  {min(widths)}px")
print(f"Maksimalna širina: {max(widths)}px")
print(f"Prosečna širina:   {int(sum(widths)/len(widths))}px")
print(f"\nMinimalna visina:  {min(heights)}px")
print(f"Maksimalna visina: {max(heights)}px")
print(f"Prosečna visina:   {int(sum(heights)/len(heights))}px")
```

```python
print("Učitavamo informacije o slikama iz skupa za validaciju:")

val_widths, val_heights = [], []

for fname in valid_files:
    img = Image.open(os.path.join(VALID_DIR, fname))
    w, h = img.size
    val_widths.append(w)
    val_heights.append(h)

print(f"\n--- VALIDATION SET ---")
print(f"Minimalna širina:  {min(val_widths)}px")
print(f"Maksimalna širina: {max(val_widths)}px")
print(f"Prosečna širina:   {int(sum(val_widths)/len(val_widths))}px")
print(f"\nMinimalna visina:  {min(val_heights)}px")
print(f"Maksimalna visina: {max(val_heights)}px")
print(f"Prosečna visina:   {int(sum(val_heights)/len(val_heights))}px")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Distribucija dimenzija slika', fontsize=16)

# Train širine
axes[0, 0].hist(widths, bins=30, color='steelblue', edgecolor='white')
axes[0, 0].set_title('Train — širina (px)')
axes[0, 0].set_xlabel('Širina (px)')
axes[0, 0].set_ylabel('Broj slika')

# Train visine
axes[0, 1].hist(heights, bins=30, color='steelblue', edgecolor='white')
axes[0, 1].set_title('Train — visina (px)')
axes[0, 1].set_xlabel('Visina (px)')
axes[0, 1].set_ylabel('Broj slika')

# Valid širine
axes[1, 0].hist(val_widths, bins=30, color='coral', edgecolor='white')
axes[1, 0].set_title('Validation — širina (px)')
axes[1, 0].set_xlabel('Širina (px)')
axes[1, 0].set_ylabel('Broj slika')

# Valid visine
axes[1, 1].hist(val_heights, bins=30, color='coral', edgecolor='white')
axes[1, 1].set_title('Validation — visina (px)')
axes[1, 1].set_xlabel('Visina (px)')
axes[1, 1].set_ylabel('Broj slika')

plt.tight_layout()
plt.show()
```
